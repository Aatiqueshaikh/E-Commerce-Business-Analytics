```python
import pandas as pd
import numpy as np

file_path = "../data/cleaned/ecommerce_cleaned_mysql.csv"

df = pd.read_csv(file_path)

df["Order Date"] = pd.to_datetime(df["Order Date"], format="mixed")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="mixed")

df["Sales"] = (
    df["Sales"].astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

df["Profit"] = (
    df["Profit"].astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

df["Quantity"] = pd.to_numeric(df["Quantity"])

df["Discount"] = (
    df["Discount"].astype(str)
    .str.replace("%", "", regex=False)
    .astype(float) / 100
)

print("Business analysis dataset loaded successfully.")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Overall Business Performance

total_revenue = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
total_units = df["Quantity"].sum()
profit_margin = (total_profit / total_revenue) * 100
average_order_value = total_revenue / total_orders

print("OVERALL BUSINESS PERFORMANCE")
print("-" * 35)
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Total Units Sold: {total_units:,}")
print(f"Profit Margin: {profit_margin:.2f}%")
print(f"Average Order Value: ${average_order_value:,.2f}")
```

```python
# Category Performance

category_analysis = (
    df.groupby("Category")
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Units=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
)

category_analysis["Profit Margin %"] = (
    category_analysis["Profit"] / category_analysis["Revenue"] * 100
)

category_analysis = category_analysis.sort_values(
    "Revenue", ascending=False
)

category_analysis
```

```python
# Regional Performance

region_analysis = (
    df.groupby("Region")
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Units=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
)

region_analysis["Profit Margin %"] = (
    region_analysis["Profit"] / region_analysis["Revenue"] * 100
)

region_analysis = region_analysis.sort_values(
    "Revenue", ascending=False
)

region_analysis
```

```python
# Sub-Category Profitability

subcategory_analysis = (
    df.groupby("Sub-Category")
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Units=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
)

subcategory_analysis["Profit Margin %"] = (
    subcategory_analysis["Profit"] / subcategory_analysis["Revenue"] * 100
)

subcategory_analysis = subcategory_analysis.sort_values(
    "Profit", ascending=False
)

subcategory_analysis
```

```python
# Yearly Business Performance

yearly_analysis = (
    df.groupby(df["Order Date"].dt.year)
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Units=("Quantity", "sum"),
        Orders=("Order ID", "nunique"),
        Customers=("Customer ID", "nunique")
    )
)

yearly_analysis["Profit Margin %"] = (
    yearly_analysis["Profit"] / yearly_analysis["Revenue"] * 100
)

yearly_analysis
```

```python
# Discount Impact Analysis

discount_analysis = (
    df.groupby("Discount")
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
)

discount_analysis["Profit Margin %"] = (
    discount_analysis["Profit"] / discount_analysis["Revenue"] * 100
)

discount_analysis
```

```python
# Top & Bottom Products by Profit

product_analysis = (
    df.groupby("Product Name")
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Units=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
)

product_analysis["Profit Margin %"] = (
    product_analysis["Profit"] / product_analysis["Revenue"] * 100
)

top_products_profit = product_analysis.sort_values(
    "Profit", ascending=False
).head(10)

bottom_products_profit = product_analysis.sort_values(
    "Profit", ascending=True
).head(10)

print("TOP 10 PRODUCTS BY PROFIT")
display(top_products_profit)

print("\nBOTTOM 10 PRODUCTS BY PROFIT")
display(bottom_products_profit)
```

```python
# Business Insights Summary

print("BUSINESS INSIGHTS")
print("=" * 50)

print("\n1. Overall Performance")
print(
    f"The business generated ${total_revenue:,.2f} in revenue "
    f"and ${total_profit:,.2f} in profit, with a "
    f"{profit_margin:.2f}% profit margin."
)

print("\n2. Yearly Growth")
print(
    f"Revenue increased from "
    f"${yearly_analysis.iloc[0]['Revenue']:,.2f} in "
    f"{yearly_analysis.index[0]} to "
    f"${yearly_analysis.iloc[-1]['Revenue']:,.2f} in "
    f"{yearly_analysis.index[-1]}."
)

print("\n3. Category Performance")
best_category = category_analysis["Profit"].idxmax()
print(
    f"{best_category} generated the highest total profit "
    f"among the three categories."
)

print("\n4. Regional Performance")
best_region = region_analysis["Profit"].idxmax()
print(
    f"{best_region} generated the highest total profit "
    f"among the four regions."
)

print("\n5. Sub-Category Performance")
best_subcategory = subcategory_analysis["Profit"].idxmax()
worst_subcategory = subcategory_analysis["Profit"].idxmin()

print(
    f"{best_subcategory} generated the highest total profit, "
    f"while {worst_subcategory} recorded the lowest total profit."
)

print("\n6. Discount Impact")
negative_discount_levels = discount_analysis[
    discount_analysis["Profit"] < 0
].index.tolist()

print(
    f"Discount levels with negative total profit: "
    f"{negative_discount_levels}"
)

print("\n7. Product Profitability")
print(
    f"The most profitable product was "
    f"{top_products_profit.index[0]}."
)

print(
    f"The largest loss-making product was "
    f"{bottom_products_profit.index[0]}."
)
```

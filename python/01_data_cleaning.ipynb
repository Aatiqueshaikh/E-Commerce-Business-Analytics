```python
import pandas as pd
import numpy as np

print("Python environment is ready.")
```

```python
# Load the cleaned e-commerce dataset

file_path = "../data/cleaned/ecommerce_cleaned_mysql.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Inspect dataset structure

print("Column Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
# Convert date columns with mixed date formats

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="mixed",
    dayfirst=False
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    format="mixed",
    dayfirst=False
)

# Convert currency columns to numeric

df["Sales"] = (
    df["Sales"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

df["Profit"] = (
    df["Profit"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

# Convert discount percentage to decimal

df["Discount"] = (
    df["Discount"]
    .astype(str)
    .str.replace("%", "", regex=False)
    .astype(float) / 100
)

print("Date and numeric columns converted successfully.")
```

```python
# Verify data types after conversion

print(df.dtypes)
```

```python
# Check for missing values

missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Check for duplicate records

duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
# Generate a statistical summary of numeric columns

df.describe()
```

```python
# Validate key numeric ranges

print("Sales <= 0:", (df["Sales"] <= 0).sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("Discount outside 0-1:", ((df["Discount"] < 0) | (df["Discount"] > 1)).sum())
print("Profit is missing:", df["Profit"].isnull().sum())
```
